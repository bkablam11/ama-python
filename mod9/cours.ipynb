```python
## POO

# Dataset 
## Deux types d'attributs :: Attribut de classe -- Attribut d'instance (objet) 

class Dataset:
  nombre_datasets_crees = 0  # attribut de classe
  nom = "Pierre"

  def __init__(self, nom, donnees):
    self.nom = nom
    self.donnees = donnees  # attribut d'instance - liste de dictionnaire 
    Dataset.nombre_datasets_crees += 1 

  def nombre_observations(self):
    return len(self.donnees)

  def change_nom(self, nouveau_nom):
    Dataset.nom = nouveau_nom


d1 = Dataset("clients", [{"age": 25, "revenu": 2200}, {"age": 41, "revenu": 3100}])
d2 = Dataset("ventes", [{"produit": "A", "quantite": 10}])


# print(Dataset.__dict__)
# print(d1.__dict__)

print(d1.nombre_observations(), d2.nombre_observations())
print("Datasets crée au total :", Dataset.nombre_datasets_crees)

print("Nom de base :", Dataset.nom) # <-- 1:

d1.change_nom("Mathieu")
print("Nom après modification d1 :", Dataset.nom) # <-- 2:
print(d1.__dict__)
d2.change_nom("Judas")
print("Nom après modification d2 :", Dataset.nom) # <-- 3:
print(d2.__dict__)


## Richnel --  Pierre - Mathieu - Judas
## Amen -- Pierre - Mathieu - Judas 



```

```python
class DatasetCorrige: 
  colonnes_obligatoires = []
  
  def __init__(self):
    self.colonnes_obligatoires = []

  def exiger_colonne(self, nom):
    self.colonnes_obligatoires.append(nom)


d_a = DatasetCorrige()
d_b = DatasetCorrige()

d_a.exiger_colonne("age")
print("Colonnes exigees par d_b :", d_b.colonnes_obligatoires)
```
