```python
## afficher la liste, longueur, equivalent, récupérer un élément de liste

## liste AMA sera une liste spéciale qui précise le type de ces éléments 
## on partira du principe que tous les items de la liste auront le même type ...

## __init__ | __repr__ | __len__ | __getitem__ | __iter__ | __eq__ | __hash__

class ListAMA: 
  
  def __init__(self, items):
    self.items = items
    self.list_type = type(items[0]) if len(items) > 0 else str

  def ajouter(self, item):
    self.items.append(item) if type(item) == self.list_type else None

  def __repr__(self):
    return f"Liste AMA (items={self.items}, type={self.list_type})"

  def __len__(self):
    return len(self.items) + 1


l1 = ListAMA(["hello"])
l2 = ListAMA([1,2,3])

```

```python
class Dataset:
  def __init__(self, nom, donnees):
    self.nom = nom
    self.donnees = donnees 

  def __repr__(self):
    return f"Dataset(nom={self.nom}, n={len(self.donnees)})"

  def __len__(self):
    return len(self.donnees)

  def __getitem__(self, indice):
    return self.donnees[indice]

  def __eq__(self, autre):
    return isinstance(autre, Dataset) and self.donnees == autre.donnees

  __hash__ = None

  def colonnes(self):
    return set(self.donnees[0].keys()) if self.donnees else set()

  def __contains__(self, nom_colonne):
    return nom_colonne in self.colonnes()


d = Dataset("clients", [{"age": 25, "revenu": 2200}])
print(d.colonnes())
print("age" in d)
print("ville" in d)
```

### Encapsulation et visibilite

```python
## @property 
# public, usage interne -- name mangling 

# toto <- public
# _toto <- usage interne  
# __toto <-- renommer l'attribut : _NomDeClasse__toto dans __dict__ 

```

```python
class Dataset:
  def __init__(self, nom, donnees):
    self.nom = nom
    self.donnees = donnees
    self.__colonne_cible = None # name mangling <-- usage interne "fort"

  @property
  def colonne_cible(self):
    return self.__colonne_cible

  @colonne_cible.setter 
  def colonne_cible(self, nom_colonne):
    colonnes_disponibles = set(self.donnees[0].keys()) if self.donnees else set()
    if nom_colonne not in colonnes_disponibles:
      raise ValueError(f"Colonne cible invalide : {nom_colonne} n'existe pas.")
    self.__colonne_cible = nom_colonne

  @property
  def taux_valeurs_manquantes(self):
    total = sum(len(obs) for obs in self.donnees)
    manquantes = sum(1 for obs in self.donnees for v in obs.values() if v is None)
    return manquantes / total if total else 0.0
    


d = Dataset("client", [{"age": 25, "revenu": None}, {"age": 41, "revenu": 3000}])

d.colonne_cible = "revenu"

print(d.colonne_cible)
print(round(d.taux_valeurs_manquantes, 2))

try:
  d.colonne_cible = "ville"
except ValueError as e:
  print(f"Erreur : {e}")

try:
  d.taux_valeurs_manquantes = 0.5
except AttributeError as e:
  print(f"Erreur : {e}")


```

```python
class Dataset:
  def __init__(self, nom, donnees):
    self.nom = nom
    self.donnees = donnees

  @property
  def donnees(self):
    return self._donnees

  # set la valeur --> self.donnees : [{"age": 12}] <-- écraser cette valeur 
  # si ceci est la nouvelle valeur : [{"age": 12, "poids": 58}, {"age": 23, "poids": 80}, {"age": 12}]
  
  @donnees.setter
  def donnees(self, valeur):
    if valeur:
      cles_reference = set(valeur[0].keys()) #  ("age", "poids")
      for observation in valeur:
        current_keys = set(observation.keys())
        if current_keys != cles_reference:
          print(f"Observation invalide : {cles_reference} (clés : {current_keys})")
          raise ValueError("Toutes les observations doivent avoir les mêmes colonnes")
    self._donnees = valeur
    

d = Dataset("clients", [{"age": 25, "revenu": 2200}, {"age": 41, "revenu": 3000}])

try:
  d.donnees = [{"age": 25}, {"age": 41, "revenu": 3000}]
except ValueError as e:
  print("Erreur : ", e)      
```

```python
class Modele:
  def entrainer(self, dataset):
    raise NotImplementedError("Les sous-classes doivent implémenter entrainer()")

  def predire(self, observation):
    raise NotImplementedError("Les sous-classes doivent implémenter predire()")


class RegressionLineaire(Modele):
  def entrainer(self, dataset):
    self.moyenne_cible = sum(obs[dataset.colonne_cible] for obs in dataset.donnees) / len(dataset.donnees)

  def predire(self, observation):
    return self.moyenne_cible # version simplifiee 


class Journalisable:
  def entrainer(self, dataset):
    print(f"[log] debut entrainement sur {len(dataset.donnees)} observations")
    super().entrainer(dataset)
    print("[log] entrainement terminé")


class RegressionJournalisee(Journalisable, RegressionLineaire):
  pass


print(RegressionJournalisee.__mro__)
```

```python
#  [{"age": 25, "revenu": 2200}, {"age": 41, "revenu": 3000}, {"age": 25, "revenu": 1200}]]
from collections import Counter

class ModeleBaseline(Modele):
  def entrainer(self, dataset):
    valeurs = [obs[dataset.colonne_cible] for obs in dataset.donnees] # <-- liste d'éléments [25, 41, 25]
    self.valeur_frequente = Counter(valeurs).most_common(1)[0][0]

  def predire(self, observation):
    return self.valeur_frequente

class BaselineJournalisee(Journalisable, ModeleBaseline):
  pass

print(BaselineJournalisee.__mro__)

d = Dataset("clients", [{"age": 25, "revenu": 12000}, {"age": 41, "revenu": 12000}, {"age": 25, "revenu": 2200}, {"age": 25, "revenu": 2200}])
d.colonne_cible = "revenu"

m = BaselineJournalisee()
m.entrainer(d)
```

### Polymorphisme et surchage 

```python
class Pipeline:
  def __init__(self, modeles):
    self.modeles = modeles # liste d'objet <-- class ...

  def entrainer_tous(self, dataset):
    for modele in self.modeles:
      modele.entrainer(dataset) 

  def comparer_predictions(self, observation):
    return {type(m).__name__: m.predire(observation) for m in self.modeles}

d = Dataset("clients", [{"age": 25, "revenu": 2200}, {"age": 41, "revenu": 3100}])
d.colonne_cible = "revenu"

pipeline = Pipeline([RegressionLineaire(), ModeleBaseline()])
pipeline.entrainer_tous(d)

print(pipeline.comparer_predictions({"age": 30}))
```

1. Implementez une classe `ModeleAleatoire(Modele)` qui, a l'entrainement, memorise

   une au hasard (module `random`).
2. Ajoutez cette instance a la liste passee au `Pipeline` deja ecrit plus haut, sans

```python
import random

class ModeleAleatoire(Modele):
  def entrainer(self, dataset):
    self.valeurs_observees = [obs[dataset.colonne_cible] for obs in dataset.donnees]

  def predire(self, observation):
    return random.choice(self.valeurs_observees)


pipeline = Pipeline([RegressionLineaire(), ModeleBaseline(), ModeleAleatoire()])
pipeline.entrainer_tous(d)
print(pipeline.comparer_predictions({"age": 30}))
```

```python
class Preprocesseur:
  def transformer(self, dataset):
    raise NotImplementedError

class NormalisationMinMax(Preprocesseur):
  def transformer(self, dataset):
    cle = dataset.colonne_cible
    valeurs = [obs[cle] for obs in dataset.donnees]
    mini, maxi = min(valeurs), max(valeurs)
    for obs in dataset.donnees:
      obs[cle] = (obs[cle] - mini) / (maxi - mini) if maxi != mini else 0.0
    return dataset

class PipelineComplet:
  def __init__(self, preprocesseur, modeles):
    self.preprocesseur = preprocesseur # "un preprocesseur"
    self.modeles = modeles 

  def executer(self, dataset):
    dataset = self.preprocesseur.transformer(dataset)
    for modele in self.modeles:
      modele.entrainer(dataset)
    return dataset


d = Dataset("clients", [{"age": 25, "revenu": 2200}, {"age": 41, "revenu": 3100}])
d.colonne_cible = "revenu"

pipeline = PipelineComplet(NormalisationMinMax(), [RegressionLineaire(), ModeleBaseline()])
pipeline.executer(d)

print(d.donnees)
    
```

### Exercice 6 -- Fil rouge, synthese finale : assembler le pipeline complet

En reutilisant toutes les briques construites depuis le debut du cours (`Dataset` avec ses
proprietes protegees, `Modele`/`RegressionLineaire`/`ModeleBaseline`/`ModeleAleatoire`,
`Journalisable`, `Preprocesseur`/`NormalisationMinMax`, `PipelineComplet`) :

1. Ajoutez une deuxieme strategie de pretraitement `StandardisationZScore` (soustraire la
   moyenne, diviser par l'ecart-type), respectant la meme interface `Preprocesseur`.
2. Construisez un `PipelineComplet` avec cette nouvelle strategie et les trois modeles vus
   en cours, sans modifier la classe `PipelineComplet`.
3. En une phrase de commentaire chacun, identifiez a quel concept du cours correspond
   chaque choix de conception fait ici (property, duck typing, composition, mixin).


```python
class StandardisationZScore(Preprocesseur):
  def transformer(self, dataset):
    cle = dataset.colonne_cible
    valeurs = [obs[cle] for obs in dataset.donnees]
    moyenne = sum(valeurs) / len(valeurs)
    ecart_type = (sum((v - moyenne)**2  for v in valeurs) / len(valeurs)) ** 0.5
    for obs in dataset.donnees: 
      obs[cle] = (obs[cle] - moyenne) / ecart_type if ecart_type else 0.0
    return dataset

d = Dataset("clients", [{"age": 25, "revenu": 2200}, {"age": 41, "revenu": 3100}, {"age": 33, "revenu": 2800}])
d.colonne_cible = "revenu"

pipeline = PipelineComplet(StandardisationZScore(), [RegressionLineaire(), ModeleBaseline(), ModeleAleatoire()])
pipeline.executer(d)

print(d.donnees)
```
