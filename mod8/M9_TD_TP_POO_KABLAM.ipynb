## Exercice 1.3 — Le constructeur alternatif : `@classmethod`

Vos donnees arrivent en dictionnaires sales. Plutot que de nettoyer avant d'appeler `Demande(...)`, on donne a la classe une **deuxieme facon de se construire**.

Ecrivez `Demande.depuis_brut(brut)` qui prend un dictionnaire brut et renvoie une `Demande`.


````python
class Demande:
    DELAI_CIBLE = 15

    def __init__(self, identifiant, service, commune, statut, delai, satisfaction=None):
        self.identifiant = identifiant
        self.service = service.strip()
        self.commune = commune.strip().title()
        self.statut = statut.strip().lower()
        self.delai = int(delai)
        self.satisfaction = satisfaction

    # TODO : @classmethod depuis_brut(cls, brut)
    #        satisfaction absente ou vide -> None
    #        pensez a utiliser cls(...) et non Demande(...)
    @classmethod # methode speciale de gestion des cas particuliers
    def depuis_brut(cls, brut):
        satisfaction = brut.get("satisfaction", "")
        satisfaction = int(satisfaction) if str(satisfaction).strip() else None
        return cls(
            brut["id"],
            brut["service"],
            brut["commune"],
            brut["statut"],
            brut["delai"],
            satisfaction,
        )

    def est_traitee(self): return self.statut == "traitee"
    def qualifier(self):
        if self.delai <= 5: return "rapide"
        elif self.delai <= 14: return "normal"
        return "lent"
    def hors_delai(self): return self.delai > self.DELAI_CIBLE


d = Demande.depuis_brut(demandes_brutes[4])   # D-005 : pas de cle satisfaction
print(d.identifiant, d.commune, d.satisfaction)
````

# BLOC 2 — Methodes speciales et encapsulation
**~30 min** · Objectif : faire en sorte que vos objets se comportent comme des objets Python natifs.

Les **methodes speciales** (ou *dunder*, pour *double underscore*) sont le contrat entre votre classe et le langage. Vous ne les appelez jamais directement : c'est Python qui les appelle quand vous ecrivez `print(d)`, `d1 == d2`, `sorted(...)`, `len(lot)`, `for x in lot`.

C'est ce qui fait la difference entre une classe qui *stocke* des donnees et une classe qui *s'integre* au langage.


## Exercice 2.1 — `__repr__` et `__str__`

Les deux affichent l'objet, mais pas pour le meme public.

- `__repr__` : pour **vous**, au debogage. Sans ambiguite, si possible re-executable.
- `__str__` : pour **l'utilisateur**, lisible.

Si `__str__` est absent, Python se rabat sur `__repr__`. L'inverse n'est pas vrai — donc **si vous n'en ecrivez qu'un seul, ecrivez `__repr__`**.


````python
class Demande:
    DELAI_CIBLE = 15

    def __init__(self, identifiant, service, commune, statut, delai, satisfaction=None):
        self.identifiant = identifiant
        self.service = service.strip()
        self.commune = commune.strip().title()
        self.statut = statut.strip().lower()
        self.delai = int(delai)
        self.satisfaction = satisfaction

    @classmethod
    def depuis_brut(cls, brut):
        sat = brut.get("satisfaction", "")
        return cls(
            brut["id"],
            brut["service"],
            brut["commune"],
            brut["statut"],
            brut["delai"],
            int(sat) if str(sat).strip() else None,
        )

    def qualifier(self):
        if self.delai <= 5: return "rapide"
        elif self.delai <= 14: return "normal"
        return "lent"

    # TODO : __repr__ -> Demande('D-001', 'Acte de naissance', delai=3)
    def __repr__(self):
        return f"Demande({self.identifiant!r}, {self.service!r}, delai={self.delai})"

    # TODO : __str__  -> D-001 · Acte de naissance (Cotonou) — 3 j [rapide]
    def __str__(self):
        return f"{self.identifiant} · {self.service} ({self.commune}) — {self.delai} j [{self.qualifier()}]"


d = Demande.depuis_brut(demandes_brutes[0])
print(repr(d))
print(str(d))
print(d)           # print appelle __str__
[d]                # une liste affiche le __repr__ de ses elements
````

## Exercice 2.2 — `__eq__`, `__hash__` et `__lt__`

Deux demandes sont **la meme** si elles ont le meme identifiant. Et on doit pouvoir les trier par delai **sans passer de `key=lambda`** — souvenez-vous du module 8, exercice 4.4.


````python
# TODO : ajoutez a la classe Demande (recopiez-la ou utilisez l'astuce ci-dessous)
# __eq__(self, autre)   -> egalite sur l'identifiant, NotImplemented si autre n'est pas une Demande
# __hash__(self)        -> hash de l'identifiant (sans lui, l'objet devient non hashable !)
# __lt__(self, autre)   -> comparaison sur le delai

# Astuce pour ne pas tout recopier : on peut greffer des methodes apres coup.
def __eq__(self, autre):
    if not isinstance(autre, Demande):
        return NotImplemented
    return self.identifiant == autre.identifiant

def __hash__(self):
    return hash(self.identifiant)

def __lt__(self, autre):
    if not isinstance(autre, Demande):
        return NotImplemented
    return self.delai < autre.delai

# Demande.__eq__ = __eq__   (decommentez quand c'est ecrit)
Demande.__eq__ = __eq__
Demande.__hash__ = __hash__
Demande.__lt__ = __lt__

a = Demande.depuis_brut(demandes_brutes[0])
b = Demande.depuis_brut(demandes_brutes[0])
c = Demande.depuis_brut(demandes_brutes[3])
print(a == b, a is b)
print(sorted([c, a]))
print({a, b})
````

## Exercice 2.3 — Encapsulation : la propriete qui valide

Au module 8, le delai negatif etait attrape par un `raise` dans une fonction. Maintenant, **l'objet refuse lui-meme d'etre dans un etat invalide** — a la construction comme a la modification.

Rendez `delai` protege par une propriete qui refuse toute valeur negative.


````python
class Demande:
    DELAI_CIBLE = 15

    def __init__(self, identifiant, service, commune, statut, delai, satisfaction=None):
        self.identifiant = identifiant
        self.service = service.strip()
        self.commune = commune.strip().title()
        self.statut = statut.strip().lower()
        self.delai = delai            # passe par le SETTER, pas directement par _delai
        self.satisfaction = satisfaction

    @classmethod
    def depuis_brut(cls, brut):
        satisfaction = brut.get("satisfaction", "")
        satisfaction = int(satisfaction) if str(satisfaction).strip() else None
        return cls(
            brut["id"],
            brut["service"],
            brut["commune"],
            brut["statut"],
            brut["delai"],
            satisfaction,
        )

    def est_traitee(self):
        return self.statut == "traitee"

    def qualifier(self):
        if self.delai <= 5:
            return "rapide"
        if self.delai <= 14:
            return "normal"
        return "lent"

    def hors_delai(self):
        return self.delai > self.DELAI_CIBLE

    # TODO : @property delai -> renvoie self._delai
    @property # methode de classe speciale getter avec mot cle @property, permet de definir un getter pour un attribut
    def delai(self):
        return self._delai

    # TODO : @delai.setter  -> convertit en int, leve DemandeInvalide si < 0
    @delai.setter
    def delai(self, valeur):
        try:
            valeur = int(valeur)
        except (TypeError, ValueError) as erreur:
            raise DemandeInvalide("Le delai doit etre un nombre entier.") from erreur
        if valeur < 0:
            raise DemandeInvalide("Le delai ne peut pas etre negatif.")
        self._delai = valeur

    def __repr__(self):
        return f"{type(self).__name__}({self.identifiant!r}, delai={self.delai})"


d = Demande("D-001", "Acte", "cotonou", "traitee", "3")
print(d.delai)
d.delai = 10
print(d.delai)
try:
    d.delai = -1      # doit lever DemandeInvalide
except DemandeInvalide as erreur:
    print("Erreur attendue :", erreur)
````

## Exercice 2.4 — `LotDeDemandes` : une classe qui se comporte comme une collection

`nettoyer_lot()` devient une classe. Avec `__len__`, `__iter__` et `__getitem__`, elle devient utilisable avec `len()`, `for ... in` et `lot[0]`, comme une liste.


````python
class LotDeDemandes:
    """Collection de demandes valides et suivi des demandes rejetees."""

    def __init__(self, demandes=None):
        self._demandes = list(demandes) if demandes else []
        self.rejets = []

    # TODO : __len__, __iter__, __getitem__, __repr__
    def __len__(self):
        return len(self._demandes)

    def __iter__(self):
        return iter(self._demandes)

    def __getitem__(self, index):
        return self._demandes[index]

    def __repr__(self):
        return f"LotDeDemandes({self._demandes!r}, rejets={len(self.rejets)})"

    # TODO : ajouter(demande)
    def ajouter(self, demande):
        if not isinstance(demande, Demande):
            raise TypeError("Le lot accepte uniquement des objets Demande.")
        self._demandes.append(demande)

    # TODO : @classmethod depuis_brutes(cls, brutes) -> construit le lot,
    #        met les echecs dans self.rejets avec la raison
    @classmethod
    def depuis_brutes(cls, brutes):
        lot = cls()
        for brute in brutes:
            try:
                lot.ajouter(Demande.depuis_brut(brute))
            except (KeyError, TypeError, ValueError, DemandeInvalide) as erreur:
                lot.rejets.append({
                    "id": brute.get("id", "INCONNU"),
                    "raison": str(erreur),
                })
        return lot


lot = LotDeDemandes.depuis_brutes(demandes_brutes)
print(len(lot), "propres |", len(lot.rejets), "rejets")
print("Rejets :", lot.rejets)
````

## Exercice 3.3 — Heritage multiple et ordre de resolution (MRO)

Python autorise l'heritage multiple. L'usage raisonnable, c'est le **mixin** : une petite classe qui ajoute une capacite, sans etre un type a part entiere.


````python
class Exportable:
    """Mixin : ajoute une capacite d'export CSV a n'importe quelle classe."""
    def en_csv(self, separateur=";"):
        return separateur.join(str(v) for v in vars(self).values())



# TODO : creez DemandeExportable qui herite de Exportable ET de Demande
class DemandeExportable(Exportable, Demande):
    """Demande avec une capacite d'export CSV."""

# TODO : affichez son MRO et expliquez l'ordre
exportable = DemandeExportable.depuis_brut(demandes_brutes[0])
print("MRO :", [classe.__name__ for classe in DemandeExportable.__mro__])
print("CSV  :", exportable.en_csv())
````
