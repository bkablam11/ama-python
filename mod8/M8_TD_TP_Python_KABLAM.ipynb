````python
# Reponse Exercice 4.1
def nettoyer_demande(demande):
    """
    Nettoie un dictionnaire de demande individuelle.
    
    Parametre :
        demande (dict) : donnee brute avec id, service, commune, statut, delai, satisfaction
    Retourne :
        dict : dictionnaire nettoye et type correctement
    Leve :
        ValueError : si le delai n'est pas convertible ou s'il est negatif
    """
    commune = demande["commune"].strip().title()
    statut = demande["statut"].strip().lower()
    
    delai_int = int(demande["delai"])
    if delai_int < 0:
        raise ValueError(f"delai negatif ({delai_int})")
        
    sat_val = demande.get("satisfaction")
    satisfaction_propre = int(sat_val) if sat_val not in (None, "") else None
    
    return {
        "id": demande["id"],
        "service": demande["service"],
        "commune": commune,
        "statut": statut,
        "delai": delai_int,
        "satisfaction": satisfaction_propre
    }

def nettoyer_lot(demandes):
    """
    Applique le nettoyage a une liste de demandes.
    
    Retourne :
        tuple : (liste_propres, liste_rejets)
    """
    propres_liste = []
    rejets_liste = []
    for d in demandes:
        try:
            propres_liste.append(nettoyer_demande(d))
        except Exception as e:
            rejets_liste.append({"id": d.get("id", "INCONNU"), "raison": str(e)})
    return propres_liste, rejets_liste

propres, rejets = nettoyer_lot(demandes_brutes)
print("Validation :", len(propres), "propres,", len(rejets), "rejets")
````

````python
# Reponse Exercice 4.2
# 1. Arguments par defaut
def resumer(demandes, seuil=14, unite="jours"):
    nb_lentes = sum(1 for d in demandes if d["delai"] > seuil)
    return f"{nb_lentes} demande(s) au-dela de {seuil} {unite}"

print("1)", resumer(propres))
print("  ", resumer(propres, seuil=5))

# 2. *args pour nombre variable de valeurs numeriques
def moyenne(*valeurs):
    if not valeurs:
        return 0
    return sum(valeurs) / len(valeurs)

print(f"2) Moyenne de 2, 4, 6, 8 : {moyenne(2, 4, 6, 8):.2f}")

# 3. **kwargs pour arguments nommes arbitraires
def journaliser(message, **details):
    print(f"[LOG] {message}")
    for k, v in details.items():
        print(f"      - {k} : {v}")

print("3) Test journaliser :")
journaliser("Demande rejetee", ref="D-003", code_erreur="VAL_ERR", auteur="agent_1")

# 4. Piege de l'argument par defaut mutable
def fonction_piegee(x, liste=[]):
    liste.append(x)
    return liste

print("4) Piege mutable appel 1 :", fonction_piegee(1))
print("   Piege mutable appel 2 :", fonction_piegee(2), "<-- la liste a ete conservee !")

# La bonne pratique :
def fonction_propre(x, liste=None):
    if liste is None:
        liste = []
    liste.append(x)
    return liste
print("   Correction propre     :", fonction_propre(2))
````

````python
# Reponse Exercice 4.6
import statistics

def rapport(demandes, rejets=None):
    """
    Genere et affiche le tableau de bord recapitulatif des demandes de service.
    """
    if rejets is None:
        rejets = []
        
    nb_valides = len(demandes)
    nb_rejets = len(rejets)
    
    # Calcul des metriques de delai
    delais = [d["delai"] for d in demandes]
    delai_moyen = statistics.mean(delais) if delais else 0
    delai_median = statistics.median(delais) if delais else 0
    
    # Calcul de la satisfaction moyenne (en ignorant les None)
    satisfactions = [d["satisfaction"] for d in demandes if d["satisfaction"] is not None]
    sat_moyenne = statistics.mean(satisfactions) if satisfactions else 0
    
    # Agregation par statut
    comptes_statut = {}
    for d in demandes:
        st = d["statut"]
        comptes_statut[st] = comptes_statut.get(st, 0) + 1
        
    # Top 3 des delais les plus longs
    top_3 = sorted(demandes, key=lambda d: d["delai"], reverse=True)[:3]
    
    # Affichage formate
    print("=" * 54)
    print("   TABLEAU DE BORD — DEMANDES DE SERVICE")
    print("=" * 54)
    print(f"Demandes valides        : {nb_valides:3}")
    print(f"Demandes rejetees       : {nb_rejets:3}")
    print(f"Delai moyen             : {delai_moyen:.2f} j")
    print(f"Delai median            : {delai_median:.2f} j")
    print(f"Satisfaction moyenne    : {sat_moyenne:.2f} /5\n")
    
    print("Par statut :")
    for statut, count in comptes_statut.items():
        barre = "#" * count
        print(f"  {statut:14} {count}  {barre}")
    
    print("\nTop 3 des delais les plus longs :")
    for i, d in enumerate(top_3, start=1):
        print(f"  {i}. {d['id']:6} {d['service']:24} {d['delai']:2} j  ({d['commune']})")
    print("=" * 54)

# Execution
rapport(propres, rejets)
````
