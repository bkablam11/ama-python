## Variables et types de donnees approfondis

````python
age = 25
taille = 1.78
nom = "Paul"
actif = True
rien = None

for variable in (age, taille, nom, actif, rien):
    print(f"{variable} -> type = {type(variable).__name__}")

# liste les fonctions usuelles sur chaque type de variable

# int
print(abs(-age))          # valeur absolue
print(age + 5)             # addition
print(age ** 2)            # puissance
print(divmod(age, 4))      # (quotient, reste)
print(bin(age))            # conversion binaire
print(hex(age))            # conversion hexadécimale
print(str(age))            # conversion en chaîne
print(float(age))          # conversion en float

# float
print(round(taille, 1))    # arrondi
print(int(taille))         # conversion en int (tronque)
print(taille.is_integer()) # vérifie si c'est un nombre entier
print(abs(-taille))        # valeur absolue

# str
print(nom.upper())         # majuscules
print(nom.lower())         # minuscules
print(nom.strip())         # supprime espaces début/fin
print(nom.replace("a", "o")) # remplacement
print(nom.split("a"))      # découpe en liste
print(len(nom))            # longueur
print(nom.startswith("P")) # test début
print(nom.endswith("l"))   # test fin
print(nom.isalpha())       # vérifie si que des lettres
print(nom.find("a"))       # position d'une sous-chaîne
print(nom[::-1])           # inverser la chaîne

# bool
print(not actif)           # inversion
print(actif and False)     # opérateur ET
print(actif or False)      # opérateur OU
print(int(actif))          # conversion en int (0 ou 1)
print(str(actif))          # conversion en chaîne

# None
print(rien is None)        # test d'identité (méthode recommandée)
print(rien == None)        # test d'égalité (moins recommandé)
print(bool(rien))          # False
````

````python
# Egalite de valeur (==) vs identite d'objet (is)
a = [1, 2, 3]
b = [1, 2, 3]
c = a

# a et b ont le meme contenu, donc == renvoie True
print("a == b :", a == b)  # True (memes valeurs)

# a et b sont deux objets differents en memoire, meme si le contenu est identique
print("a is b :", a is b)  # False (objets differents)

# c pointe vers le meme objet que a (pas une copie)
print("a is c :", a is c)  # True (meme objet en memoire)

# Pour verifier : on peut afficher l'adresse memoire avec id() <-- 
print("id(a) :", id(a))
print("id(b) :", id(b))
print("id(c) :", id(c))
# id(a) et id(c) seront identiques, id(b) sera different

# Consequence importante : si on modifie c, ca modifie aussi a
# car ce sont le meme objet
c.append(4)
print("a apres modification de c :", a)  # [1, 2, 3, 4]
print("b n'a pas change :", b)           # [1, 2, 3]

# Regle generale :
# - "==" compare le CONTENU (valeurs egales ?)
# - "is" compare l'IDENTITE (meme objet en memoire ?)
# On utilise "is" surtout pour comparer a None, True, False
````

## Structures de donnees : listes, tuples, dictionnaires, sets

````python

# Listes : ordonnees et mutables
fruits = ["pomme", "banane", "mangue"]
fruits.append("orange")
fruits[0] = "fraise"
print("Liste :", fruits)
print("Slicing [1:3] :", fruits[1:3])

# Fonctions usuelles sur les listes
fruits.append("Python")      # ajoute a la fin
fruits.insert(1, "kiwi")     # insere a une position precise
fruits.remove("mangue")      # supprime la premiere occurrence trouvee
fruits.pop()                 # supprime et renvoie le dernier element (ou un index donne)
fruits.sort()                # trie la liste sur place (ordre croissant)
fruits.reverse()             # inverse l'ordre de la liste
print("Longueur :", len(fruits))       # nombre d'elements
print("Occurrences :", fruits.count("Python"))  # compte les occurrences
print("Index :", fruits.index("fraise") if "fraise" in fruits else "absent")  # position d'un element
print("Copie :", fruits.copy())        # copie superficielle de la liste
print("Extension :", fruits + ["cerise"])  # concatenation (ou .extend())
print("Existence :", "fraise" in fruits)   # test d'appartenance
````

````python

# Tuples : immuables
coordonnees = (48.8566, 2.3522)
print("Tuple :", coordonnees)
try:
    coordonnees[0] = 0
except TypeError as e:
    print("Erreur attendue :", e)

# Fonctions usuelles sur les tuples
print("Longueur :", len(coordonnees))          # nombre d'elements
print("Index :", coordonnees.index(2.3522))    # position d'une valeur
print("Occurrences :", coordonnees.count(48.8566))  # compte les occurrences
print("Existence :", 48.8566 in coordonnees)   # test d'appartenance
lat, lon = coordonnees                          # unpacking (destructuration)
print("Unpacking :", lat, lon)
print("Conversion en liste :", list(coordonnees))  # tuple -> liste (pour le rendre mutable)
````

````python

# Dictionnaires
etudiant = {
    "nom": "Paul",
    "cohorte": 2,
    "modules_valides": ["Module 8"]
}
etudiant["moyenne"] = 15.5
print(etudiant)

for cle, valeur in etudiant.items():
    print(f"  {cle} -> {valeur}")

# Fonctions usuelles sur les dictionnaires
print("Cles :", etudiant.keys())              # liste des cles
print("Valeurs :", etudiant.values())         # liste des valeurs
print("Get :", etudiant.get("age", "inconnu"))  # acces securise (valeur par defaut si absent)
etudiant.update({"age": 22})                  # ajoute/modifie plusieurs cles a la fois
supprime = etudiant.pop("age")                # supprime une cle et renvoie sa valeur
print("Supprime :", supprime)
print("Existence :", "nom" in etudiant)       # test d'appartenance (verifie les cles)
print("Copie :", etudiant.copy())             # copie superficielle
etudiant.setdefault("ville", "Paris")         # ajoute la cle seulement si elle n'existe pas
print(etudiant)
# etudiant.clear()                            # viderait completement le dictionnaire

````

## Structures de controle : conditions et boucles

#### Conditions
Le bloc `if / elif / else` permet d'executer du code selon une condition booleenne.

#### Boucles
- `for` : itere sur une sequence (liste, tuple, chaine, `range()`, etc.).
- `while` : repete tant qu'une condition reste vraie.
- `break` interrompt la boucle, `continue` passe a l'iteration suivante.

````python
note = 13

if note >= 16:
    mention = "Tres bien"
elif note >= 14:
    mention = "Bien"
elif note >= 10:
    mention = "Passable"
else:
    mention = "Insuffisant"

print(f"Note : {note}/20 -> {mention}")
````

````python
for i in range(5):
    print("Iteration", i)

print("---")

modules = ["Python de base", "POO", "Bibliotheques essentielles"]
for indice, module in enumerate(modules, start=8):
    print(f"Module {indice} : {module}")

print("---")

compteur = 0
while True: 
    compteur += 1   # <-- notez bien l'incrémentation ici, sinon on aurait une boucle infinie !!
    if compteur % 2 == 0:
        continue
    if compteur > 7:
        break
    print("Nombre impair :", compteur)
````

### 5. Gestion des exceptions (try / except / finally)

Une exception est une erreur detectee pendant l'execution. Plutot que de laisser le
programme planter, on peut l'intercepter avec `try / except`.

- `try` : bloc de code potentiellement risque.
- `except TypeErreur as e` : capture un type d'erreur precis (toujours preferer des
  exceptions precises a un `except:` generique).
- `else` : execute si aucune exception ne s'est produite.
- `finally` : execute dans tous les cas (nettoyage, fermeture de fichier, etc.).

````python
def diviser(a, b):
    try:
        resultat = a / b
    except ZeroDivisionError:
        print("Erreur : division par zero impossible.")
        return None
    except TypeError:
        print("Erreur : les deux arguments doivent etre des nombres.")
        return None
    else:
        print("Division reussie, aucune exception levee.")
        return resultat
    finally:
        print(f"Tentative de division de {a} par {b} terminee.\n")

print(diviser(10, 2))
print(diviser(10, 0))
print(diviser(10, "deux"))
````

````python
def inscrire_etudiant(age):
    if age < 0:
        raise ValueError("L'age ne peut pas etre negatif.")
    if age < 16:
        raise ValueError("L'etudiant doit avoir au moins 16 ans.")
    return f"Inscription validee pour un etudiant de {age} ans."

for age_test in [22, -5, 10]:
    try:
        print(inscrire_etudiant(age_test))
    except ValueError as erreur:
        print("Inscription refusee :", erreur)
````

### 6. Fonctions : definition, arguments, portee, fonctions natives, lambda

Une fonction encapsule un bloc de code reutilisable. On la definit avec `def`.

#### Types d'arguments
- Positionnels : `def f(a, b)`
- Par mot-cle : `f(a=1, b=2)`
- Valeurs par defaut : `def f(a, b=10)`
- `*args` : nombre variable d'arguments positionnels (tuple)
- `**kwargs` : nombre variable d'arguments nommes (dict)

#### Portee (scope)
- Une variable definie dans une fonction est locale : elle n'existe pas a l'exterieur.
- Le mot-cle `global` permet (rarement recommande !!!) de modifier une variable globale depuis
  une fonction.

#### Fonctions lambda
Fonctions anonymes, sur une seule ligne, utiles pour des traitements courts (souvent avec
`sorted()`, `map()`, `filter()`).

````python
def presenter(nom, ville="Paris", *diplomes, **infos):
    print(f"{nom}, base(e) a {ville}.")
    if diplomes:
        print("Diplomes :", ", ".join(diplomes))
    for cle, valeur in infos.items():
        print(f"  {cle} : {valeur}")

presenter("Paul", "Paris", "Master Informatique", "Certif ML", role="Software Engineer", entreprise="Bivariant")
````

````python
# Scope 
x = 10

def modifier_localement():
    x = 99
    print("Dans la fonction, x =", x)

def modifier_globalement():
    global x
    x = 99

modifier_localement()
print("Apres appel local, x reste =", x)

modifier_globalement()
print("Apres appel avec 'global', x devient =", x)
````

````python
notes = [12, 18, 9, 15, 20, 7]

print("len :", len(notes))
print("sum :", sum(notes))
print("min/max :", min(notes), max(notes))
print("sorted decroissant :", sorted(notes, reverse=True))

carre = lambda x: x ** 2
print("carre(5) =", carre(5))

notes_au_carre = list(map(lambda n: n ** 2, notes))
notes_admises = list(filter(lambda n: n >= 10, notes))
print("Notes au carre :", notes_au_carre)
print("Notes admises (>=10) :", notes_admises)

etudiants = [{"nom": "Paul", "note": 15}, {"nom": "Fatou", "note": 18}]
etudiants_tries = sorted(etudiants, key=lambda e: e["note"], reverse=True)
print(etudiants_tries)
````

### Comprehensions de listes et de dictionnaires

Les comprehensions offrent une syntaxe concise pour construire des listes ou dictionnaires
a partir d'une sequence, en une seule ligne.

Syntaxe generale :
```python
[expression for element in iterable if condition]      # liste
{cle: valeur for element in iterable if condition}      # dictionnaire
```

````python
carres_boucle = []
for n in range(10):
    carres_boucle.append(n ** 2)

carres_comprehension = [n ** 2 for n in range(10)]

print(carres_boucle)
print(carres_comprehension)
print("Resultats identiques :", carres_boucle == carres_comprehension)
````

````python
pairs_au_carre = [n ** 2 for n in range(10) if n % 2 == 0]
print("Carres des nombres pairs :", pairs_au_carre)

prenoms = ["Paul", "Fatou", "Koffi", "Ibrahim"]
longueurs = {prenom: len(prenom) for prenom in prenoms}
print("Longueurs des prenoms :", longueurs)

table = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print("Table :", table)
````
